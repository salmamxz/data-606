## Working with dictionaries

```python
# converting series
sales_dict = Sales.to_dict()
sales_dict
```

```python
sales_ser = pd.Series(sales_dict)
sales_ser
```

## Adding entries and working with nan null values 

```python
# We can create a new series from an already existing series 
new_sales = pd.Series(Sales, index=["Don", "Mike", "Salma", "Sally"])
new_sales
```

```python
print(np.isnan(new_sales["Sally"]))
```

```python
# nan is not same as none 
new_sales["Sally"] is None 
```

```python
# is nan can be used on the entire series 
np.isnan(new_sales)
```
