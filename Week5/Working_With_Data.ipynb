```python
# I am ranking petal length from smallest to largest.
# The smallest petal length gets rank 1.
df['petal_length_rank'] = df['petal_length'].rank(method='min', ascending=True).astype(int)

# I am sorting by the new rank to check the result.
df.sort_values('petal_length_rank').head(10)
```

```python
# I am finding the average numerical measurements for each species.
mean_by_species = df.groupby('species').mean(numeric_only=True)
mean_by_species
```

```python
# I am making a separate DataFrame for each species.
# This lets me look at each group on its own.
setosa_df = df[df['species'] == 'setosa'].copy()
versicolor_df = df[df['species'] == 'versicolor'].copy()
virginica_df = df[df['species'] == 'virginica'].copy()

print('Setosa:', setosa_df.shape)
print('Versicolor:', versicolor_df.shape)
print('Virginica:', virginica_df.shape)
```

```python
# I am creating one DataFrame with the sepal measurements for all species.
sepal_info = df[['sepal_length', 'sepal_width', 'species']].copy()
sepal_info.head()
```

```python
# I am creating another DataFrame with the petal measurements for all species.
petal_info = df[['petal_length', 'petal_width', 'species']].copy()
petal_info.head()
```

```python
# I am adding an ID for each flower so the sepal and petal details can be matched correctly.
iris_with_id = df.reset_index(names='flower_id')

sepal_info = iris_with_id[['flower_id', 'sepal_length', 'sepal_width', 'species']].copy()
petal_info = iris_with_id[['flower_id', 'petal_length', 'petal_width', 'species']].copy()

# I am merging the two DataFrames using the flower ID.
merged_df = pd.merge(
    sepal_info,
    petal_info,
    on='flower_id',
    suffixes=('_sepal', '_petal')
)

# Both DataFrames contain species, so I am dropping the repeated species column.
# The merged result keeps species_sepal and species_petal; they contain the same values.
merged_df = merged_df.drop(columns=['species_petal']).rename(columns={'species_sepal': 'species'})

merged_df.head()
```

```python
# I am joining the three species DataFrames back together.
iris_concatenated = pd.concat(
    [setosa_df, versicolor_df, virginica_df],
    ignore_index=True
)

# I am checking the result and comparing its shape with the original dataset.
print('Original shape:', df.shape)
print('Concatenated shape:', iris_concatenated.shape)
iris_concatenated.head()
```

```python
titanic_clean = titanic.copy()
```

```python
# I remove exact duplicate rows.
titanic_clean = titanic_clean.drop_duplicates()

# I check how many duplicate rows remain.
titanic_clean.duplicated().sum()
```

```python
titanic_clean.isnull().sum()
```

```python
# I fill missing ages with the median age.
titanic_clean['age'] = titanic_clean['age'].fillna(
    titanic_clean['age'].median()
)

# I fill missing embarkation ports with the most common port.
titanic_clean['embarked'] = titanic_clean['embarked'].fillna(
    titanic_clean['embarked'].mode()[0]
)

# I remove columns that I will not use in this analysis.
titanic_clean = titanic_clean.drop(columns=['deck', 'embark_town'])

# I check the remaining missing values.
titanic_clean.isnull().sum()
```

```python
# I group passengers by class and calculate their average fare and age.
class_averages = titanic_clean.groupby('class', observed=True).agg(
    average_fare=('fare', 'mean'),
    average_age=('age', 'mean')
)

class_averages
```

```python
# I calculate each passenger's fare as a percentage of all fares combined.
total_fare = titanic_clean['fare'].sum()

titanic_clean['fare_percentage'] = (
    titanic_clean['fare'] / total_fare
) * 100

# I display the fare and its percentage of the total.
titanic_clean[['fare', 'fare_percentage']].head()
```

```python
survivors_by_sex = titanic_clean[
    titanic_clean['survived'] == 1
].groupby('sex').size()

survivors_by_sex
```

```python
# I calculate the survival rate for female and male passengers.
survival_by_sex = titanic_clean.groupby('sex', observed=True).agg(
    passengers=('survived', 'count'),
    survivors=('survived', 'sum'),
    survival_rate=('survived', 'mean')
)

# I convert the survival rate into a percentage.
survival_by_sex['survival_rate'] = (
    survival_by_sex['survival_rate'] * 100
).round(2)

survival_by_sex
```

```python
survivors_by_class = titanic_clean.groupby(
    'class', observed=True
)['survived'].agg(['count', 'sum', 'mean'])

survivors_by_class.columns = [
    'passengers', 'survivors', 'survival_rate'
]

# I convert the survival rate into a percentage.
survivors_by_class['survival_rate'] = (
    survivors_by_class['survival_rate'] * 100
).round(2)

survivors_by_class
```

```python
# I create a column to separate children from adults.
titanic_clean['age_group'] = np.where(
    titanic_clean['age'] < 18,
    'Child',
    'Adult'
)

# I compare survival rates between children and adults.
age_survival = titanic_clean.groupby('age_group').agg(
    passengers=('survived', 'count'),
    survivors=('survived', 'sum'),
    survival_rate=('survived', 'mean')
)

age_survival['survival_rate'] = (
    age_survival['survival_rate'] * 100
).round(2)

age_survival
```

```python
port_survival = titanic_clean.groupby('embarked', observed=True).agg(
    passengers=('survived', 'count'),
    survivors=('survived', 'sum'),
    survival_rate=('survived', 'mean')
)

port_survival['survival_rate'] = (
    port_survival['survival_rate'] * 100
).round(2)

port_survival
```
