# NumPy Data Analysis Exercises

This notebook uses NumPy to analyse student scores, weather data and retail sales. It also explores random data generation, reshaping arrays and transposing data.

```python
import numpy as np
```

## 1. Student Scores Analysis

Create an array of student scores and calculate the highest, lowest, average and total scores. Then add five bonus marks to every student.

**New NumPy methods:**
- `np.max()` returns the largest value.
- `np.min()` returns the smallest value.
- `np.mean()` calculates the average.
- `np.sum()` adds all values.
- Array addition adds the bonus to each element.

```python
scores = np.array([78, 85, 92, 67, 88, 95, 74, 81, 90, 69])

print("Highest score:", np.max(scores))
print("Lowest score:", np.min(scores))
print("Average score:", np.mean(scores))
print("Total score:", np.sum(scores))

bonus_scores = scores + 5

print("Scores with bonus marks:", bonus_scores)
```

## 2. Weather Data

Analyse daily temperatures to identify hot and cold days, calculate the average temperature, and count the days that exceeded the average.

**New NumPy concept:** A comparison such as `temperatures > 25` returns a Boolean array. We can use this condition to select matching temperatures and `np.sum()` to count how many values meet a condition.

```python
temperatures = np.array([
    18, 21, 25, 17, 16,
    23, 28, 30, 22, 19,
    15, 24, 27, 20
])

above_25 = temperatures[temperatures > 25]
below_20 = temperatures[temperatures < 20]
average_temp = np.mean(temperatures)

print("Days above 25 degrees:", above_25)
print("Days below 20 degrees:", below_20)
print("Average temperature:", average_temp)
print("Hottest temperature:", np.max(temperatures))
print("Coldest temperature:", np.min(temperatures))
print("Days exceeding average:", np.sum(temperatures > average_temp))
```

## 3. Retail Sales Analysis

The rows represent three stores, and the columns represent four weeks of sales.

**New NumPy methods:**
- `axis=1` calculates across each row, giving a result per store.
- `axis=0` calculates down each column, giving a result per week.
- `np.argmax()` returns the index of the largest value. Python indexing starts at zero.

```python
sales = np.array([
    [120, 150, 180, 200],
    [90, 110, 140, 160],
    [200, 220, 250, 300]
])

store_totals = np.sum(sales, axis=1)
week_totals = np.sum(sales, axis=0)
overall_sales = np.sum(sales)
best_store = np.argmax(store_totals) + 1
best_week = np.argmax(week_totals) + 1
average_per_store = np.mean(sales, axis=1)

print("Total sales per store:", store_totals)
print("Total sales per week:", week_totals)
print("Overall sales:", overall_sales)
print("Best performing store:", best_store)
print("Best performing week:", best_week)
print("Average weekly sales per store:", average_per_store)
```

## 4. Random Customer Purchases

Generate ten random customer purchase amounts between £1 and £500 and store them in a NumPy array.

**New NumPy methods:**
- `np.random.default_rng()` creates a random number generator.
- `rng.integers()` generates random integers.
- `size=10` specifies how many values to generate.
- The upper limit in `rng.integers()` is excluded, so we use `501` to include £500.

```python
rng = np.random.default_rng()

purchases = rng.integers(1, 501, size=10)

print("Customer purchases (£):", purchases)
print("Total purchases (£):", np.sum(purchases))
print("Average purchase (£):", np.mean(purchases))
```

## 5. Reshape and Transpose

**`reshape()`** changes the dimensions of an array without changing its data. The new dimensions must accommodate the same number of elements.

**`transpose()`** swaps the array's axes. For a two-dimensional array, it switches rows and columns.

```python
numbers = np.array([1, 2, 3, 4, 5, 6])

reshaped = numbers.reshape(2, 3)

print("Original array:")
print(numbers)

print("Reshaped array (2 rows, 3 columns):")
print(reshaped)

print("Transposed array (3 rows, 2 columns):")
print(reshaped.transpose())
```
