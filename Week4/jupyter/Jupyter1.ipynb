```python
import numpy as np
import pandas as pd
from IPython.display import display

numbers = [10, 20, 30, 40, 50]
array = np.array(numbers)

results = pd.DataFrame({
    "Index": range(len(numbers)),
    "Python List": numbers,
    "NumPy Array": array,
    "Doubled Value": array * 2
})

display(
    results.style
    .set_caption("Python Lists vs NumPy Arrays")
    .format(precision=2)
    .background_gradient(subset=["NumPy Array"], cmap="Blues")
    .background_gradient(subset=["Doubled Value"], cmap="Greens")
)
```

```python

import numpy as np
import pandas as pd
from IPython.display import display, Markdown

books = pd.DataFrame({
    "Book": ["The Silent Patient", "It Ends with Us", "Fourth Wing",
             "The Cruel Prince", "Book Lovers", "The Love Hypothesis"],
    "Genre": ["Thriller", "Romance", "Fantasy", "Fantasy", "Romance", "Romance"],
    "Units Sold": [850, 1200, 980, 760, 690, 1100],
    "Price (£)": [9.99, 8.99, 12.99, 8.50, 9.50, 10.99]
})

books["Revenue (£)"] = books["Units Sold"] * books["Price (£)"]

display(Markdown("# Book Market Dashboard"))
display(Markdown("An overview of fictional book sales and revenue."))

display(Markdown(
    f"### Total Revenue: £{books['Revenue (£)'].sum():,.2f}  \n"
    f"**Units sold:** {books['Units Sold'].sum():,} | "
    f"**Books analysed:** {len(books)}"
))

display(
    books.style
    .format({"Price (£)": "£{:.2f}", "Revenue (£)": "£{:,.2f}"})
    .background_gradient(subset=["Units Sold"], cmap="Blues")
    .background_gradient(subset=["Revenue (£)"], cmap="Greens")
    .bar(subset=["Units Sold"], color="#E8A0BF")
    .set_properties(**{"text-align": "left"})
    .hide(axis="index")
)
```

```python

import matplotlib.pyplot as plt
import numpy as np

products = ["Books", "E-books", "Audiobooks", "Kindle"]
revenue = np.array([12500, 8200, 6400, 3900])

fig, ax = plt.subplots(figsize=(10, 5))

bars = ax.bar(
    products, revenue,
    color=["#D8A7B1", "#B8C7E0", "#B7D7C5", "#E8D6B3"],
    width=0.65
)

ax.set_title("Revenue by Product Category", fontsize=17, pad=18)
ax.set_ylabel("Revenue (£)")
ax.set_xlabel("Product Category")
ax.set_ylim(0, max(revenue) * 1.2)
ax.spines[["top", "right"]].set_visible(False)
ax.grid(axis="y", linestyle="--", alpha=0.25)
ax.set_axisbelow(True)

ax.bar_label(bars, labels=[f"£{v:,.0f}" for v in revenue], padding=5)

plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
from IPython.display import display, Markdown

students = pd.DataFrame({
    "Student": ["Student A", "Student B", "Student C", "Student D",
                "Student E", "Student F", "Student G", "Student H"],
    "Python": [78, 92, 45, 86, 67, 39, 95, 73],
    "SQL":    [82, 88, 52, 91, 71, 44, 89, 77]
})

students["Average"] = students[["Python", "SQL"]].mean(axis=1)
students["Result"] = np.where(students["Average"] >= 50, "Pass", "Needs Improvement")

pass_rate = (students["Average"] >= 50).mean() * 100

display(Markdown("# Student Performance Report"))
display(Markdown(
    f"**Average score:** {students['Average'].mean():.1f}%  \n"
    f"**Pass rate:** {pass_rate:.1f}%  \n"
    f"**Highest average:** {students.loc[students['Average'].idxmax(), 'Student']}"
))

display(
    students.style
    .format({"Python": "{:.0f}", "SQL": "{:.0f}", "Average": "{:.1f}%"})
    .background_gradient(subset=["Python", "SQL", "Average"], cmap="RdYlGn",
                         vmin=0, vmax=100)
    .hide(axis="index")
)
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Marketing Spend": [100, 150, 200, 250, 300, 350],
    "Website Visits": [1200, 1700, 2100, 2800, 3200, 3900],
    "Orders": [90, 130, 155, 220, 250, 310],
    "Returns": [5, 8, 7, 12, 10, 15]
})

correlation = data.corr(numeric_only=True)

fig, ax = plt.subplots(figsize=(8, 6))
image = ax.imshow(correlation, cmap="RdYlBu_r", vmin=-1, vmax=1)

ax.set_xticks(range(len(correlation.columns)))
ax.set_xticklabels(correlation.columns, rotation=35, ha="right")
ax.set_yticks(range(len(correlation.index)))
ax.set_yticklabels(correlation.index)

for i in range(len(correlation)):
    for j in range(len(correlation.columns)):
        ax.text(j, i, f"{correlation.iloc[i, j]:.2f}",
                ha="center", va="center", fontsize=11)

ax.set_title("Business Metrics Correlation", fontsize=16, pad=16)
fig.colorbar(image, ax=ax, label="Correlation")
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
from IPython.display import display, Markdown

np.random.seed(42)

daily_sales = np.random.randint(50, 250, size=30)
average = np.mean(daily_sales)
best_day = np.argmax(daily_sales) + 1
worst_day = np.argmin(daily_sales) + 1
above_average = daily_sales[daily_sales > average]

display(Markdown("# 30-Day Sales Challenge"))

metrics = pd.DataFrame({
    "Metric": [
        "Total Sales",
        "Daily Average",
        "Highest Sales",
        "Lowest Sales",
        "Median Sales",
        "Days Above Average"
    ],
    "Value": [
        np.sum(daily_sales),
        average,
        np.max(daily_sales),
        np.min(daily_sales),
        np.median(daily_sales),
        len(above_average)
    ]
})

display(
    metrics.style
    .format({"Value": "{:,.2f}"})
    .background_gradient(subset=["Value"], cmap="PuBu")
    .hide(axis="index")
)

display(Markdown(
    f"**Best day:** Day {best_day}  \n"
    f"**Lowest day:** Day {worst_day}  \n"
    f"**Days above average:** {len(above_average)} out of 30"
))
```
