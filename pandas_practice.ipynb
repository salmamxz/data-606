## 1. Create a DataFrame

A DataFrame stores data in rows and columns, similar to a spreadsheet or database table.

```python
data = {
    "Name": ["Amira", "Adam", "Sara", "Youssef", "Lina"],
    "Age": [22, 25, 21, 28, 24],
    "Country": ["UK", "France", "UK", "Spain", "France"],
    "Score": [85, 78, 92, 67, 88]
}

df = pd.DataFrame(data)

display(df)
```

## 2. Exploring a DataFrame

Pandas provides methods and attributes to help us understand the structure and contents of a dataset.

```python
print("First five rows:")
display(df.head())
```

```python
print("DataFrame dimensions (rows, columns):")
print(df.shape)
```

```python
print("Column names:")
print(df.columns)

```

```python
print("Data types and non-null values:")
df.info()
```

```python
print("Summary statistics:")
display(df.describe())
```

## 3. Selecting Data

Pandas allows us to select individual columns, multiple columns and particular rows.

```python
print("Select one column:")
display(df["Name"])
```

```python
print("Select multiple columns:")
display(df[["Name", "Score"]])
```

```python
print("Select the first two rows:")
display(df.iloc[0:2])
```

```python

print("Select the first row using its position:")
display(df.iloc[0])
```

```python
print("Select rows where the score is above 80:")
display(df[df["Score"] > 80])
```

```python
print("Select rows using a condition:")
display(df.loc[df["Country"] == "UK"])
```

## 4. Adding Data

We can add a new column or add a new row to an existing DataFrame.

```python
df["Passed"] = df["Score"] >= 70

df.loc[len(df)] = {
    "Name": "Omar",
    "Age": 23,
    "Country": "UK",
    "Score": 74,
    "Passed": True
}

display(df)
```

## 5. Statistical Analysis

Pandas provides functions to calculate totals, averages, minimums, maximums and other useful statistics.

```python
print("Average age:", df["Age"].mean())
print("Average score:", df["Score"].mean())
print("Highest score:", df["Score"].max())
print("Lowest score:", df["Score"].min())
print("Total scores:", df["Score"].sum())
print("Number of records:", df["Name"].count())

print("Number of people in each country:")
display(df["Country"].value_counts())
```

## 6. Working with CSV Files

CSV stands for Comma-Separated Values. It is a common file format for storing tabular data.

Pandas uses `to_csv()` to save data and `read_csv()` to load it.

```python
df.to_csv("customers.csv", index=False)

customers_df = pd.read_csv("customers.csv")

display(customers_df)
```

## 7. Data Cleaning

Real datasets can contain missing values, duplicate rows and incorrect data types. Pandas provides methods to identify and correct these problems.

```python
messy_data = {
    "Name": ["Amira", "Adam", "Sara", "Adam", None],
    "Age": [22, 25, None, 25, 30],
    "Score": [85, 78, 92, 78, None]
}

messy_df = pd.DataFrame(messy_data)

display(messy_df)
```

```python
print("Missing values per column:")
print(messy_df.isnull().sum())

print("Number of duplicate rows:")
print(messy_df.duplicated().sum())

clean_df = messy_df.drop_duplicates().copy()

clean_df["Age"] = clean_df["Age"].fillna(clean_df["Age"].mean())
clean_df["Score"] = clean_df["Score"].fillna(clean_df["Score"].mean())
clean_df["Name"] = clean_df["Name"].fillna("Unknown")

print("Cleaned DataFrame:")
display(clean_df)

print("Missing values after cleaning:")
print(clean_df.isnull().sum())
```
